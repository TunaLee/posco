## 1. 값과 변수

**실습.** 파이썬이 다루는 값의 종류를 확인한다. `type()` 이 자료형을 알려 준다.

```python
print(22,      type(22))
print(7.25,    type(7.25))
print("B00115", type("B00115"))
print(True,    type(True))
print([22, 38], type([22, 38]))
```

**실습.** 따옴표 안은 전부 글자다. 같은 `+` 라도 자료형에 따라 하는 일이 다르다.

```python
print(1234 + 5)        # 덧셈
print("1234" + "5")    # 이어 붙이기
print("3" * 2)         # 두 번 반복
```

**실습.** 나눗셈이 세 가지다. `/` 는 나누어 떨어져도 실수를 준다.

```python
print(7 / 2)    # 나누기 — 항상 float
print(7 // 2)   # 몫만
print(7 % 2)    # 나머지
print(2 ** 10)  # 거듭제곱
```

**실습.** 문자열도 순서를 가진다. 인덱스는 0부터, 마지막은 -1이다.

```python
batch = "B00115"
print(batch[0], batch[-1])
print(batch[2:5])       # 2 이상 5 미만
print(len(batch))
print(batch.lower())
print(batch.startswith("B"))
```

### 불변과 가변

**실습.** `b = a` 는 복사가 아니다. 이름을 하나 더 붙일 뿐이다. 두 결과를 비교한다.

```python
a = 10
b = a
b = b + 5
print("불변:", a, b)      # 값이 갈라진다

a = [1, 2]
b = a
b.append(3)
print("가변:", a, b)      # 같이 바뀐다
print("같은 객체인가:", id(a) == id(b))
```

**실습.** `.copy()` 는 얕은 복사다. 안쪽에 또 자료 구조가 있으면 그건 공유된다.

```python
import copy

rows = [{"line": "A", "temp": 898}, {"line": "E", "temp": 829}]

shallow = rows.copy()
shallow[0]["age"] = 99
print("얕은 복사 후 원본:", rows[0]["age"])   # 99 — 원본도 바뀐다

rows[0]["age"] = 22                            # 되돌리고
deep = copy.deepcopy(rows)
deep[0]["age"] = 99
print("깊은 복사 후 원본:", rows[0]["age"])   # 22 — 그대로
```

## 3. 흐름 제어

**실습.** 비교식의 결과는 True / False 두 값뿐이다. 이것이 조건문의 입력이 된다.

```python
age = 22
print(age < 18, type(age < 18))
print("Male" == "male")          # 대소문자를 구분한다
print(18 <= age < 65)            # 범위는 이어 쓴다
print(age > 20 and age < 30)
print(age < 18 or age > 60)
print(not (age < 18))
```

**실습.** 0, 빈 문자열, 빈 리스트는 거짓으로 취급된다.

```python
for value in [0, 0.0, "", [], {}, None, 1, "a", [0]]:
    print(repr(value), "→", bool(value))
```

**실습.** 조건이 셋 이상이면 elif 로 잇는다. 위에서부터 걸리는 첫 줄만 실행된다.

```python
def group_of(age):
    if age < 13:
        return "어린이"
    elif age < 20:
        return "청소년"
    elif age < 65:
        return "성인"
    return "노년"

for a in [7, 15, 42, 70]:
    print(a, group_of(a))
```

**실습.** 반복문은 꺼낼 대상이 있을 때 for, 끝을 조건이 정할 때 while 이다.

```python
temps = [898, 835, 829]

for t in temps:
    print("온도:", t)

for i, t in enumerate(temps, start=1):
    print(i, t)

for i in range(1, 6):
    print("*" * i)
```

**실습.** break 는 반복 자체를 끝내고, continue 는 이번 회차만 건너뛴다.

```python
temps = [898, 835, 760, 860]

for t in temps:
    if t < 800:
        print("저온 발견:", t)
        break
    print("확인:", t)

print("---")
for t in temps:
    if t < 800:
        continue
    print("집계:", t)
```

### 컴프리헨션

**실습.** 반복문을 한 줄로 접는다. 괄호가 결과 자료형을 정한다.

```python
lines = ["A", "C", "E"]
temps = [898, 835, 829]

print([L.lower() for L in lines])                        # list
print({L: t for L, t in zip(lines, temps)})              # dict
print({t // 10 * 10 for t in temps})                     # set
print(sum(t for t in temps))                             # generator
print([t for t in temps if t >= 850])                    # 조건으로 거르기
```

## 4. 함수

**실습.** 내장 함수는 가져오지 않고 바로 쓴다.

```python
temps = [898, 835, 829, 860]

print(len(temps), max(temps), min(temps), sum(temps))
print(sum(temps) / len(temps))     # 평균 — 내장 함수에는 없다
print(sorted(temps, reverse=True))
```

**실습.** def 로 만든 함수는 호출해야 실행된다. return 이 값을 돌려준다.

```python
def age_group(age, adult=18):
    if age < adult:
        return "미성년"
    return "성인"

print(age_group(12))
print(age_group(12, adult=10))    # 기본값을 바꿔 부른다
```

**실습.** print 만 하는 함수는 값을 돌려주지 않는다. 그 결과를 계산에 쓰면 None 이 흐른다.

```python
def show(a):
    print("성인" if a >= 18 else "미성년")

def judge(a):
    return "성인" if a >= 18 else "미성년"

r1 = show(22)
r2 = judge(22)
print("show 의 결과:", r1)     # None
print("judge 의 결과:", r2)
```

**실습.** 함수 안에서 만든 이름은 함수가 끝나면 사라진다. 가변 인자는 밖까지 바뀐다.

```python
def add_one(n):
    n = n + 1
    return n

def append_one(lst):
    lst.append(1)

x = 10
add_one(x)
print("불변 인자:", x)          # 10 — 안 바뀐다

nums = [0]
append_one(nums)
print("가변 인자:", nums)       # [0, 1] — 바뀐다
```
