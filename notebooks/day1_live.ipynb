## 1. 값과 변수

> **실습문제 1.** `n` 을 정수로 바꿔 `10` 을 더한 값을 `total` 에 담는다.

```python
n = "32"
total = int(n) + 10

assert total == 42, f"기대 42, 실제 {total}"
print("통과")
```

> **실습문제 2.** `origin` 을 건드리지 않고 값을 추가한 새 리스트를 `copy_` 에 만든다.

```python
origin = [1, 2, 3]
copy_ = origin.copy()
copy_.append(99)

assert origin == [1, 2, 3], f"원본이 바뀌었다: {origin}"
assert copy_ == [1, 2, 3, 99], f"기대 [1,2,3,99], 실제 {copy_}"
print("통과")
```

## 2. 자료 구조

> **실습문제 3.** `scores` 에서 가장 큰 값과 가장 작은 값의 차이를 `gap` 에 담는다.

```python
scores = [88, 95, 72, 100, 64]
gap = max(scores) - min(scores)

assert gap == 36, f"기대 36, 실제 {gap}"
print("통과")
```

> **실습문제 4.** `config` 에서 은닉층 목록의 **첫 번째 값**을 `first` 에 담는다.

```python
config = {"lr": 0.001, "epochs": 50, "hidden": [64, 32]}
first = config["hidden"][0]

assert first == 64, f"기대 64, 실제 {first}"
print("통과")
```

## 3. 흐름 제어

> **실습문제 5.** 온도가 820 이상 880 미만이면 True 가 되도록 `in_range` 를 만든다.

```python
temp = 855
in_range = 820 <= temp < 880

assert in_range is True, f"기대 True, 실제 {in_range}"
temp = 910
assert not (820 <= temp < 880), "조건식을 다시 확인한다"
print("통과")
```

> **실습문제 6.** 1부터 100까지 더한 값을 `total` 에 담는다. `range` 를 쓴다.

```python
total = 0
for i in range(1, 101):
    total = total + i

assert total == 5050, f"기대 5050, 실제 {total}"
print("통과")
```

> **실습문제 7.** `nums` 의 각 값을 제곱한 리스트를 컴프리헨션으로 `squares` 에 담는다.

```python
nums = [1, 2, 3, 4, 5]
squares = [n ** 2 for n in nums]

assert squares == [1, 4, 9, 16, 25], f"기대 [1,4,9,16,25], 실제 {squares}"
print("통과")
```

## 4. 함수

> **실습문제 8.** 섭씨를 화씨로 바꾸는 함수를 만든다. 공식은 `F = C * 9/5 + 32` 다.

```python
def to_fahrenheit(c):
    return c * 9 / 5 + 32

assert to_fahrenheit(0) == 32, f"0도 → 기대 32, 실제 {to_fahrenheit(0)}"
assert to_fahrenheit(100) == 212, f"100도 → 기대 212, 실제 {to_fahrenheit(100)}"
print("통과")
```
