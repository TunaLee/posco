> **실습문제 16.** 숫자 목록을 받아 **최댓값과 최솟값을 뺀 나머지의 평균**을 돌려주는 함수를 만든다.
> 값이 두 개 이하면 `0` 을 돌려준다.

```python
def trimmed_mean(values):
    if len(values) <= 2:
        return 0
    rest = sorted(values)[1:-1]
    return sum(rest) / len(rest)
assert trimmed_mean([1, 5, 6, 7, 100]) == 6, f"실제 {trimmed_mean([1,5,6,7,100])}"
assert trimmed_mean([3, 9]) == 0, "값이 둘이면 0 이어야 한다"
print("통과")
```

> **빈칸 문제 18.** 점수 목록을 받아 `{"A": n, "B": n, ...}` 형태로 등급 분포를 돌려주는 함수를 만든다.
> 90↑ A · 80↑ B · 70↑ C · 나머지 D. 한 명도 없는 등급은 키에 넣지 않는다.

```python
def grade_counts(scores):
    result = {}
    for s in scores:
        if s >= 90:
            g = 'A'
        elif s >= 80:
            g = 'B'
        elif s >= 70:
            g = 'C'
        else:
            g = 'D'
        result[g] = result.get(g, 0) + 1
    return result

out = grade_counts([95, 82, 71, 64, 88, 91])
assert out == {"A": 2, "B": 2, "C": 1, "D": 1}, f"실제 {out}"
assert grade_counts([]) == {}, "빈 목록은 빈 딕셔너리여야 한다"
print("통과")
```

> **실습문제 18.** 배치 목록을 받아 **교대조별 양품률 딕셔너리**를 돌려주는 함수를 만든다.
> 결과는 `{"day": 1.0, "night": 0.666…}` 형태다.

```python
batches = [
    {"설비호기": "A", "교대조": "day",   "양품여부": 1},
    {"설비호기": "C", "교대조": "night", "양품여부": 0},
    {"설비호기": "E", "교대조": "night", "양품여부": 1},
    {"설비호기": "B", "교대조": "day",   "양품여부": 1},
    {"설비호기": "D", "교대조": "night", "양품여부": 1},
]
def pass_rate_by_shift(rows):
    result = {}
    for sh in set(r["교대조"] for r in rows):
        group = [r for r in rows if r["교대조"] == sh]
        result[sh] = sum(r["양품여부"] for r in group) / len(group)
    return result
out = pass_rate_by_shift(batches)
assert set(out.keys()) == {"day", "night"}, f"키가 다르다: {out.keys()}"
assert out["day"] == 1.0, f"day 기대 1.0, 실제 {out['day']}"
assert abs(out["night"] - 2/3) < 1e-9, f"night 기대 0.666…, 실제 {out['night']}"
print("통과")
```

> **실습문제 19.** 온도 목록을 받아 **10도 구간별 배치 수**를 돌려주는 함수를 만든다.
> `[898, 835, 829, 860]` → `{890: 1, 830: 1, 820: 1, 860: 1}`. 결측(`None`)은 건너뛴다.

```python
def by_band(temps):
    result = {}
    for t in temps:
        if t is None:
            continue
        b = t // 10 * 10
        result[b] = result.get(b, 0) + 1
    return result
assert by_band([898, 835, 829, 860]) == {890: 1, 830: 1, 820: 1, 860: 1}, f"실제 {by_band([898,835,829,860])}"
assert by_band([898, None, 895]) == {890: 2}, f"결측 제외 후 기대 {{890: 2}}, 실제 {by_band([898, None, 895])}"
print("통과")
```
