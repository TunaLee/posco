> **문제 5.** `origin` 을 건드리지 않고 값을 추가한 새 리스트를 `copy_` 에 만든다.

```python
origin = [1, 2, 3]
copy_ = origin.copy()
copy_.append(99)

assert origin == [1, 2, 3], f"원본이 바뀌었다: {origin}"
assert copy_ == [1, 2, 3, 99], f"기대 [1,2,3,99], 실제 {copy_}"
print("통과")
```

**실습.** `.copy()` 는 얕은 복사다. 안쪽에 또 자료 구조가 있으면 그건 공유된다.

```python
import copy

rows = [{"name": "Owen", "age": 22}, {"name": "Laina", "age": 26}]

shallow = rows.copy()
shallow[0]["age"] = 99
print("얕은 복사 후 원본:", rows[0]["age"])   # 99 — 원본도 바뀐다

rows[0]["age"] = 22                            # 되돌리고
deep = copy.deepcopy(rows)
deep[0]["age"] = 99
print("깊은 복사 후 원본:", rows[0]["age"])   # 22 — 그대로
```

```python
def mean(values):
    if not values:
        return 0
    return sum(values) / len(values)

assert mean([2, 4, 6]) == 4, f"기대 4, 실제 {mean([2,4,6])}"
assert mean([]) == 0, f"빈 리스트 → 기대 0, 실제 {mean([])}"
print("통과")
```

> **문제 23.** 승객 목록에서 **여성 생존율**을 구해 `rate` 에 담는다. 소수 그대로 둔다.

> **문제 25.** 단어 목록에서 **각 단어가 몇 번 나왔는지** 세어 딕셔너리 `counts` 를 만든다.

```python
words = ["red", "blue", "red", "green", "blue", "red"]
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1

assert counts == {"red": 3, "blue": 2, "green": 1}, f"기대 {{'red':3,'blue':2,'green':1}}, 실제 {counts}"
print("통과")
```

> **문제 26.** 점수 목록을 받아 `{"A": n, "B": n, ...}` 형태로 등급 분포를 돌려주는 함수를 만든다.
> 90↑ A · 80↑ B · 70↑ C · 나머지 D. 한 명도 없는 등급은 키에 넣지 않는다.

```python
def grade_counts(scores):
    result = {}
    for s in scores:
        if s >= 90:
            g = 'A'
        elif s >= 80:
            g = 'B'
        elif s >= 70:
            g = 'C'
        else:
            g = 'D'
        result[g] = result.get(g, 0) + 1
    return result

out = grade_counts([95, 82, 71, 64, 88, 91])
assert out == {"A": 2, "B": 2, "C": 1, "D": 1}, f"실제 {out}"
assert grade_counts([]) == {}, "빈 목록은 빈 딕셔너리여야 한다"
print("통과")
```
