> **실습문제 8.** 숫자 목록을 받아 **최댓값과 최솟값을 뺀 나머지의 평균**을 돌려주는 함수를 만든다.
> 값이 두 개 이하면 `0` 을 돌려준다.

```python
def trimmed_mean(values):
    if len(values) <= 2:
        return 0
    rest = sorted(values)[1:-1]
    return sum(rest) / len(rest)
assert trimmed_mean([1, 5, 6, 7, 100]) == 6, f"실제 {trimmed_mean([1,5,6,7,100])}"
assert trimmed_mean([3, 9]) == 0, "값이 둘이면 0 이어야 한다"
print("통과")
```

> 공백과 대소문자는 무시한다. `"Never odd or even"` 은 True 다.

```python
def is_palindrome(text):
    s = text.replace(" ", "").lower()
    return s == s[::-1]
assert is_palindrome("Never odd or even") is True, "True 여야 한다"
assert is_palindrome("hello") is False, "False 여야 한다"
print("통과")
```

> **빈칸 문제 26.** 점수 목록을 받아 `{"A": n, "B": n, ...}` 형태로 등급 분포를 돌려주는 함수를 만든다.
> 90↑ A · 80↑ B · 70↑ C · 나머지 D. 한 명도 없는 등급은 키에 넣지 않는다.

```python
def grade_counts(scores):
    result = {}
    for s in scores:
        if s >= 90:
            g = 'A'
        elif s >= 80:
            g = 'B'
        elif s >= 70:
            g = 'C'
        else:
            g = 'D'
        result[g] = result.get(g, 0) + 1
    return result

out = grade_counts([95, 82, 71, 64, 88, 91])
assert out == {"A": 2, "B": 2, "C": 1, "D": 1}, f"실제 {out}"
assert grade_counts([]) == {}, "빈 목록은 빈 딕셔너리여야 한다"
print("통과")
```

> **실습문제 10.** 승객 목록을 받아 **성별 생존율 딕셔너리**를 돌려주는 함수를 만든다.
> 결과는 `{"male": 0.0, "female": 0.666…}` 형태다.

```python
passengers = [
    {"name": "Owen",     "sex": "male",   "survived": 0},
    {"name": "Florence", "sex": "female", "survived": 1},
    {"name": "Laina",    "sex": "female", "survived": 1},
    {"name": "William",  "sex": "male",   "survived": 0},
    {"name": "Lily",     "sex": "female", "survived": 0},
]
def survival_by_sex(rows):
    result = {}
    for sex in set(r["sex"] for r in rows):
        group = [r for r in rows if r["sex"] == sex]
        result[sex] = sum(r["survived"] for r in group) / len(group)
    return result
out = survival_by_sex(passengers)
assert set(out.keys()) == {"male", "female"}, f"키가 다르다: {out.keys()}"
assert out["male"] == 0.0, f"male 기대 0.0, 실제 {out['male']}"
assert abs(out["female"] - 2/3) < 1e-9, f"female 기대 0.666…, 실제 {out['female']}"
print("통과")
```

> **실습문제 11.** 나이 목록을 받아 **연령대별 인원수**를 돌려주는 함수를 만든다.
> `[22, 38, 26, 35, 41]` → `{20: 2, 30: 2, 40: 1}`. 결측(`None`)은 건너뛴다.

```python
def by_decade(ages):
    result = {}
    for a in ages:
        if a is None:
            continue
        d = a // 10 * 10
        result[d] = result.get(d, 0) + 1
    return result
assert by_decade([22, 38, 26, 35, 41]) == {20: 2, 30: 2, 40: 1}, f"실제 {by_decade([22,38,26,35,41])}"
assert by_decade([22, None, 25]) == {20: 3} or by_decade([22, None, 25]) == {20: 2}, "결측을 건너뛰어야 한다"
assert by_decade([22, None, 25]) == {20: 2}, f"결측 제외 후 기대 {{20: 2}}, 실제 {by_decade([22, None, 25])}"
print("통과")
```
