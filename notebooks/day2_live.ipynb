## 0. NumPy

> **실습문제 1.** `temps` 를 배열로 만들고 전부 화씨로 바꿔 `f` 에 담는다. 화씨 = 섭씨 × 9/5 + 32.

```python
import numpy as np
temps = [898, 835, 829]
c = np.array(temps)
f = c * 9 / 5 + 32

assert list(f.round(1)) == [1648.4, 1535.0, 1524.2], f'실제 {list(f.round(1))}'
print('통과')
```

> **실습문제 2.** `X` 를 6행짜리로 바꿔 `Y` 에 담는다. 열 수는 `-1` 로 맡긴다.

```python
import numpy as np
X = np.arange(24)
Y = X.reshape(6, -1)

assert Y.shape == (6, 4), f'기대 (6, 4), 실제 {Y.shape}'
print('통과')
```

## 1. Pandas — 읽고 확인하기

> **실습문제 3.** `df` 에서 **행 수**를 `n_rows`, **열 수**를 `n_cols` 에 담는다.

```python
import pandas as pd
df = pd.read_csv('https://tunalee.github.io/posco/data/batch_quality.csv', thousands=',', na_values=['N/A', '-'])
df['설비호기'] = df['설비호기'].str.strip().str.upper()
n_rows = df.shape[0]
n_cols = df.shape[1]

assert (n_rows, n_cols) == (1412, 11), f'실제 {(n_rows, n_cols)}'
print('통과')
```

## 2. Pandas — 고르고 계산하기

> **실습문제 4.** 불량(`양품여부 == 0`)인 행만 골라 `bad` 에 담는다.

```python
import pandas as pd
df = pd.read_csv('https://tunalee.github.io/posco/data/batch_quality.csv', thousands=',', na_values=['N/A', '-'])
df['설비호기'] = df['설비호기'].str.strip().str.upper()
bad = df[df['양품여부'] == 0]

assert len(bad) == 265, f'기대 265, 실제 {len(bad)}'
print('통과')
```

## 3. Pandas — 묶고 합치기

> **실습문제 5.** 설비별 **불량률**을 구해 `bad_rate` 에 담는다. 불량률 = 1 − 양품률.

```python
import pandas as pd
df = pd.read_csv('https://tunalee.github.io/posco/data/batch_quality.csv', thousands=',', na_values=['N/A', '-'])
df['설비호기'] = df['설비호기'].str.strip().str.upper()
bad_rate = 1 - df.groupby('설비호기')['양품여부'].mean()

assert abs(bad_rate['A'] - 0.221) < 0.002, f"A 실제 {bad_rate['A']}"
assert bad_rate.idxmax() == 'A', f'가장 나쁜 설비는 A 여야 한다: {bad_rate.idxmax()}'
print('통과')
```

## 4. 전처리

> **실습문제 6.** `수분율` 결측을 **중앙값**으로 채우고, 채운 흔적을 `moisture_missing` 열에 남긴다.

```python
import pandas as pd
df = pd.read_csv('https://tunalee.github.io/posco/data/batch_quality.csv', thousands=',', na_values=['N/A', '-'])
df['설비호기'] = df['설비호기'].str.strip().str.upper()
df['moisture_missing'] = df['수분율'].isna()
df['수분율'] = df['수분율'].fillna(df['수분율'].median())

assert df['수분율'].isna().sum() == 0, '아직 결측이 남았다'
assert df['moisture_missing'].sum() == 73, f"기대 73, 실제 {df['moisture_missing'].sum()}"
print('통과')
```

## 5. 시각화

> **실습문제 7.** 설비별 **평균 용량**을 막대 그래프로 그린다.

```python
import pandas as pd
df = pd.read_csv('https://tunalee.github.io/posco/data/batch_quality.csv', thousands=',', na_values=['N/A', '-'])
df['설비호기'] = df['설비호기'].str.strip().str.upper()
import matplotlib.pyplot as plt
ax = df.groupby('설비호기')['방전용량'].mean().plot.bar()
plt.show()

assert ax is not None, '그래프 객체가 없다'
print('통과')
```
